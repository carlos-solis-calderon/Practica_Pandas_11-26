```python
import pandas as pd
#traer el DataSet
df = pd.read_csv('../raw/streaming.csv')



```

```python
#exploracion del dataset

print("Filas y columnas:", df.shape)
print("Nombres de columnas:", df.columns.tolist())
df.head()
df.info()
df.describe()
```

```python
#ver duplicados y nulos
print("Valores nulos por columna:")
print(df.isnull().sum())
print("\nFilas duplicadas:", df.duplicated().sum())
```

#Exploración realizada, comenzamos la limpieza del dataset


```python
df = df.drop(columns=['Session_ID', 'Device_ID', 'Location', 'Playback_Quality', 'Languages', 'Interaction_Events'])
```

Comprobar el borrado

```python
print("Filas y columnas:", df.shape)
print("Columnas:", df.columns.to_list())
df.head()
df.info()
```

Comenzar a filtrar por lo que nos piden

```python
#genero favorito de los hombres

hombres = df[df['Gender'] == 'Male']


genero_preferido = hombres.groupby('Genre').agg(
    Total_Minutos=('Duration_Watched (minutes)', 'sum')
).sort_values(by='Total_Minutos', ascending=False)

print("El género preferido de los hombres es:", genero_preferido.index[0])


```

```python
#corroborar si los smartphones son el dispositivo en el cual permanecen mas tiempo los usuarios viendo streaming

tiempo_por_dispositivo = df.groupby('Device_Type').agg(
        total_tiempo =('Duration_Watched (minutes)', 'sum')
).sort_values(by='total_tiempo', ascending=False)

print("El dispositivo mas utilizado es:", tiempo_por_dispositivo.index[0])
```

```python
#edades que suelen pagar mas premium

df_premium = df[df['Subscription_Status'] == 'Premium']

edades_premium = df_premium.groupby('Age').agg(
    total_usuarios_premium=('Subscription_Status', 'count')
).sort_values(by='total_usuarios_premium', ascending=False)

print("La edad que más paga Premium es:", edades_premium.index[0])
print("La segunda edad que más paga Premium es:", edades_premium.index[1])
print("La tercera que más paga Premium es:", edades_premium.index[2])
print("La cuarta edad que más paga Premium es:", edades_premium.index[3])


```

```python
#video que mas tiempo de visualizacion tiene

video_mayor_visualizacion = df.groupby('Video_ID').agg(
    video_mas_tiempo = ('Duration_Watched (minutes)', 'sum')
).sort_values(by='video_mas_tiempo', ascending=False)

print("El video que mas tiempo de visualizacion es el", video_mayor_visualizacion.index[0])
```

```python
#El género que más le gustó a las mujeres (rating)

df_mujeres = df[df['Gender'] == 'Female']

genero_favorito_mujeres = (
    df_mujeres.groupby('Genre')
    .agg(promedio_rating=('Ratings', 'mean'))
    .sort_values(by='promedio_rating', ascending=False)
)

print(
    'El género que más le gustó a las mujeres es:',
    genero_favorito_mujeres.index[0],
)
```
